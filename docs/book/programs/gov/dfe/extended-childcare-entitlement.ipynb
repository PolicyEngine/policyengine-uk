# Extended childcare entitlement

[Official government guidance](https://www.gov.uk/check-eligible-free-childcare-if-youre-working?step-by-step-nav=f517cd57-3c18-4bb9-aa8b-1b907e279bf9) | [Legislation](https://www.legislation.gov.uk/uksi/2017/1160/pdfs/uksiem_20171160_en.pdf)

## Program description

### Overview

For eligible working parents in England, the combined funded childcare entitlement follows this rollout:

* From **April 2024**, **2-year-olds** receive **15 hours** per week.
* From **September 2024**, children aged **9 months to 2 years** receive **15 hours** per week.
* From **September 2025**, children aged **9 months to 2 years** receive **30 hours** per week.
* Children aged **3 to 4 years** receive **30 hours** per week throughout these periods.

The younger-child rollout is set out in the [2023 amending regulations, regulations 1(2) and 2(5), (8)](https://www.legislation.gov.uk/uksi/2023/1330/made/data.html). Hours are funded over 38 weeks, or stretched over more weeks without increasing the annual total.

The 30 hours include any universal (3- and 4-year-old) or targeted (2-year-old) hours available to the child. [Childcare Act 2016 s.1(6)](https://www.legislation.gov.uk/ukpga/2016/5/section/1) counts childcare available under the 2006 Act towards the total; [2014 regulations 3–4](https://www.legislation.gov.uk/uksi/2014/2147/made/data.html) establish those base entitlements. DfE's [statutory guidance A1.11](https://www.gov.uk/government/publications/early-education-and-childcare--2/early-education-and-childcare-valid-from-1-april-2026#section-a1-eligibility) requires the first 15 hours for a child eligible for both targeted and working-parent provision to be funded as targeted hours. Qualifying for the working-parent entitlement never removes universal or targeted eligibility.

A child in a state-funded or council-funded reception class gets no funded hours outside its school place ([statutory guidance](https://www.gov.uk/government/publications/early-education-and-childcare--2/early-education-and-childcare-valid-from-1-april-2026), footnotes 7 and 17, paras A4.8 and C.6), so it gets neither universal nor working-parent hours once `current_education` is primary or a later stage (`early_years_child_in_school`).

### Eligibility requirements

#### Work status
Eligible individuals must be either employed or about to start employment. This includes those on qualifying leave types: **sick**, **annual**, **shared parental**, **maternity**, **paternity**, or **adoption** leave.

Non-working partners may qualify if they receive: **incapacity benefit**, **severe disablement allowance**, **carer's allowance**, **limited capability for work benefit**, or **contribution-based employment and support allowance**.

#### Income requirements
Minimum earnings thresholds (per **3 months**) vary by age:

| Age group | Minimum earnings | Weekly equivalent |
|-----------|-----------------|-------------------|
| **21+** | **£2,380** | **£183** |
| **18-20** | **£1,788** | **£137** |
| **Under 18/Apprentices** | **£1,331** | **£102** |

These thresholds are based on national minimum wage calculations for **16 hours** per week.

Maximum income limit: Adjusted net income must not exceed **£100,000** per year.

### Additional rules

#### Income assessment
The program offers flexible income assessment:
* Combines **multiple income sources**
* Allows **annual income averaging**
* Gives **self-employed** individuals choice between total or self-employment income

Certain types of income are excluded from calculations: **dividends**, **interest**, **property income**, and **pension payments**.

* Converts yearly income to quarterly periods
* Applies minimum wage thresholds based on work hours

### 5. Entitlement calculation
Final calculations in [`extended_childcare_entitlement_per_child.py`](https://github.com/PolicyEngine/policyengine-uk/blob/main/policyengine_uk/variables/gov/dfe/extended_childcare_entitlement/extended_childcare_entitlement_per_child.py), summed for the family by [`extended_childcare_entitlement.py`](https://github.com/PolicyEngine/policyengine-uk/blob/main/policyengine_uk/variables/gov/dfe/extended_childcare_entitlement/extended_childcare_entitlement.py):
* Determines weekly hours based on child's age
* Limits them to the hours the child and the family use (`max_free_entitlement_hours_used`, `maximum_extended_childcare_hours_usage`)
* Applies appropriate funding rates
* For children participating in working-parent provision, allocates their first funded hours, up to 15 per week, to the applicable universal or targeted entitlement based on legal eligibility, regardless of the standalone take-up flags
* Allocates only the remaining funded hours to `extended_childcare_entitlement_per_child`; a child with a universal or targeted entitlement who uses 15 hours a week or fewer has all funded hours allocated to that base programme
* Preserves combined support as the greater of the claimed standalone base entitlement and the working-parent total after usage caps, rather than subtracting an unpaid base award
* Sums the children in the benefit unit

The annual model evaluates parameters at January and does not prorate the April or September expansions within a calendar year.

| 2         | 15                | 0                      | £8.28       | £4,719.6         |
| 3         | 30                | 15                     | £5.88       | £3,351.6         |
